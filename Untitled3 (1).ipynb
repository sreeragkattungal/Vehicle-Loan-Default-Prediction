```python
import pandas as pd
import numpy as np

# ---------------------------------------------------------
# 1. Standardize column names
# ---------------------------------------------------------

df.columns = (
    df.columns
      .str.strip()
      .str.replace(".", "_", regex=False)
      .str.replace(" ", "_", regex=False)
)

print("Columns after standardization:")
print(df.columns.tolist())


# ---------------------------------------------------------
# 2. Convert AVERAGE_ACCT_AGE and CREDIT_HISTORY_LENGTH
#    Example: "1yrs 11mon" -> 23 months
# ---------------------------------------------------------

def convert_to_months(value):
    if pd.isna(value):
        return np.nan

    value = str(value).strip()

    # Extract years and months
    match = value.lower().replace(" ", "").strip()

    years = 0
    months = 0

    # Find years
    if "yrs" in match:
        years_part = match.split("yrs")[0]
        years = int(years_part) if years_part.isdigit() else 0

    # Find months
    if "mon" in match:
        months_part = match.split("mon")[0].split("yrs")[-1]
        months = int(months_part) if months_part.isdigit() else 0

    return years * 12 + months


df["AVERAGE_ACCT_AGE_MONTHS"] = df["AVERAGE_ACCT_AGE"].apply(convert_to_months)
df["CREDIT_HISTORY_LENGTH_MONTHS"] = df["CREDIT_HISTORY_LENGTH"].apply(convert_to_months)

# Drop the original string columns
df.drop(columns=["AVERAGE_ACCT_AGE", "CREDIT_HISTORY_LENGTH"], inplace=True)


# ---------------------------------------------------------
# 3. Parse Date.of.Birth and DisbursalDate
# ---------------------------------------------------------

df["Date_of_Birth"] = pd.to_datetime(
    df["Date_of_Birth"],
    dayfirst=True,
    errors="coerce"
)

df["DisbursalDate"] = pd.to_datetime(
    df["DisbursalDate"],
    dayfirst=True,
    errors="coerce"
)


# ---------------------------------------------------------
# 4. Fix DOB century issue
# ---------------------------------------------------------
# In this dataset, DOBs parsed as years >= 2000 are treated
# as a two-digit-year century parsing issue and shifted back
# by 100 years.

mask = df["Date_of_Birth"].dt.year >= 2000

df.loc[mask, "Date_of_Birth"] = (
    df.loc[mask, "Date_of_Birth"] - pd.DateOffset(years=100)
)


# ---------------------------------------------------------
# 5. Calculate age at disbursal
# ---------------------------------------------------------

df["Age_at_Disbursal"] = (
    df["DisbursalDate"].dt.year
    - df["Date_of_Birth"].dt.year
    - (
        (df["DisbursalDate"].dt.month < df["Date_of_Birth"].dt.month)
        |
        (
            (df["DisbursalDate"].dt.month == df["Date_of_Birth"].dt.month)
            & (df["DisbursalDate"].dt.day < df["Date_of_Birth"].dt.day)
        )
    ).astype(int)
)


# ---------------------------------------------------------
# 6. Impute missing Employment.Type
# ---------------------------------------------------------

df["Employment_Type"] = df["Employment_Type"].fillna("Unknown")


# ---------------------------------------------------------
# 7. Drop high-cardinality ID columns
# ---------------------------------------------------------

id_columns = [
    "UniqueID",
    "Employee_code_ID",
    "Current_pincode_ID"
]

df.drop(columns=id_columns, inplace=True)


# ---------------------------------------------------------
# 8. Check the cleaned dataset
# ---------------------------------------------------------

print("\nShape after cleaning:", df.shape)

print("\nMissing values:")
print(df.isnull().sum().sort_values(ascending=False).head(15))

print("\nData types:")
print(df.dtypes)

print("\nFirst 5 rows:")
display(df.head())
```

```python
import re
import pandas as pd
import numpy as np

# Safely load the original raw column values if they aren't present in df
if "AVERAGE_ACCT_AGE" not in df.columns or "CREDIT_HISTORY_LENGTH" not in df.columns:
    original_df = pd.read_csv("/content/train.csv")

    if "AVERAGE_ACCT_AGE" not in df.columns:
        df["AVERAGE_ACCT_AGE"] = original_df["AVERAGE.ACCT.AGE"]

    if "CREDIT_HISTORY_LENGTH" not in df.columns:
        df["CREDIT_HISTORY_LENGTH"] = original_df["CREDIT.HISTORY.LENGTH"]

def convert_to_months(value):
    if pd.isna(value):
        return np.nan

    value = str(value).strip().lower()

    years = re.search(r'(\d+)\s*yrs?', value)
    months = re.search(r'(\d+)\s*mon', value)

    years = int(years.group(1)) if years else 0
    months = int(months.group(1)) if months else 0

    return years * 12 + months

# Apply the transformation
df["AVERAGE_ACCT_AGE_MONTHS"] = df["AVERAGE_ACCT_AGE"].apply(convert_to_months)
df["CREDIT_HISTORY_LENGTH_MONTHS"] = df["CREDIT_HISTORY_LENGTH"].apply(convert_to_months)

# Drop original columns
df.drop(
    columns=["AVERAGE_ACCT_AGE", "CREDIT_HISTORY_LENGTH"],
    inplace=True,
    errors="ignore"
)

print("Successfully parsed columns!")
print(df[["AVERAGE_ACCT_AGE_MONTHS", "CREDIT_HISTORY_LENGTH_MONTHS"]].head())
```

```python
# 1. Fix age anomalies
# Row-wise operation: no parameter is learned from the dataset.

def fix_age(part):
    part = part.copy()

    # Parse DOB and DisbursalDate explicitly with dayfirst
    part["date_of_birth"] = pd.to_datetime(
        part["date_of_birth"],
        errors="coerce"
    )

    part["disbursaldate"] = pd.to_datetime(
        part["disbursaldate"],
        errors="coerce"
    )

    # Calculate initial age
    part["age_at_disbursal"] = (
        part["disbursaldate"].dt.year
        - part["date_of_birth"].dt.year
        - (
            (part["disbursaldate"].dt.month < part["date_of_birth"].dt.month)
            |
            (
                (part["disbursaldate"].dt.month == part["date_of_birth"].dt.month)
                &
                (part["disbursaldate"].dt.day < part["date_of_birth"].dt.day)
            )
        ).astype(int)
    )

    # Fix century issue (if birth year comes after disbursal year)
    bad_dob = (
        part["date_of_birth"].notna()
        & part["disbursaldate"].notna()
        & (part["date_of_birth"].dt.year > part["disbursaldate"].dt.year)
    )

    print("DOB century wrap corrected:", bad_dob.sum())

    part.loc[bad_dob, "date_of_birth"] = (
        part.loc[bad_dob, "date_of_birth"]
        - pd.DateOffset(years=100)
    )

    # Recalculate age
    part["age_at_disbursal"] = (
        part["disbursaldate"].dt.year
        - part["date_of_birth"].dt.year
        - (
            (part["disbursaldate"].dt.month < part["date_of_birth"].dt.month)
            |
            (
                (part["disbursaldate"].dt.month == part["date_of_birth"].dt.month)
                &
                (part["disbursaldate"].dt.day < part["date_of_birth"].dt.day)
            )
        ).astype(int)
    )

    # Treat ages outside plausible range (under 18 or over 75) as missing (NaN)
    implausible_age = (
        (part["age_at_disbursal"] < 18)
        | (part["age_at_disbursal"] > 75)
    )
    print("Implausible ages set to NaN:", implausible_age.sum())

    part.loc[implausible_age, "age_at_disbursal"] = np.nan

    return part

train_df = fix_age(train_df)
val_df = fix_age(val_df)

if heldout_test_df is not None:
    heldout_test_df = fix_age(heldout_test_df)

print("\nAge summary after fixing:")
print(train_df["age_at_disbursal"].describe())
```

```python
# 2. Collapse secondary-account columns

def add_secondary_totals(part):
    part = part.copy()

    # Convert relevant columns to numeric
    account_pairs = {
        "total_no_of_accts": ("pri_no_of_accts", "sec_no_of_accts"),
        "total_active_accts": ("pri_active_accts", "sec_active_accts"),
        "total_overdue_accts": ("pri_overdue_accts", "sec_overdue_accts"),
        "total_current_balance": ("pri_current_balance", "sec_current_balance"),
        "total_sanctioned_amount": ("pri_sanctioned_amount", "sec_sanctioned_amount"),
        "total_disbursed_amount": ("pri_disbursed_amount", "sec_disbursed_amount"),
        "total_instal_amt": ("primary_instal_amt", "sec_instal_amt")
    }

    for new_col, (pri_col, sec_col) in account_pairs.items():
        part[pri_col] = pd.to_numeric(part[pri_col], errors="coerce")
        part[sec_col] = pd.to_numeric(part[sec_col], errors="coerce")

        part[new_col] = (
            part[pri_col].fillna(0)
            + part[sec_col].fillna(0)
        )

    # Secondary account indicator
    part["has_secondary_account"] = (
        part["sec_no_of_accts"].fillna(0) > 0
    ).astype(int)

    # Drop raw secondary columns
    secondary_cols = [
        "sec_no_of_accts",
        "sec_active_accts",
        "sec_overdue_accts",
        "sec_current_balance",
        "sec_sanctioned_amount",
        "sec_disbursed_amount",
        "sec_instal_amt"
    ]

    part.drop(
        columns=[c for c in secondary_cols if c in part.columns],
        inplace=True
    )

    return part


train_df = add_secondary_totals(train_df)
val_df = add_secondary_totals(val_df)

if heldout_test_df is not None:
    heldout_test_df = add_secondary_totals(heldout_test_df)

print("Secondary-account consolidation completed.")

display(
    train_df[
        [
            "total_no_of_accts",
            "total_active_accts",
            "total_overdue_accts",
            "total_current_balance",
            "total_sanctioned_amount",
            "total_disbursed_amount",
            "total_instal_amt",
            "has_secondary_account"
        ]
    ].head()
)
```

```python
# 3. Resolve collinear pairs

# Compute 99th percentile of the ratio strictly on train_df first to avoid leakage
temp_denominator = train_df["total_sanctioned_amount"].replace(0, np.nan)
temp_ratio = train_df["total_disbursed_amount"] / temp_denominator
RATIO_CAP = temp_ratio.dropna().quantile(0.99)

def resolve_collinearity(part, ratio_cap):
    part = part.copy()

    # 1. Create binary flag for "sanctioned = 0"
    part["sanctioned_amount_is_zero"] = (
        (part["total_sanctioned_amount"].fillna(0) == 0)
    ).astype(int)

    # 2. Disbursed / sanctioned ratio with zero denominator mapped to NaN
    denominator = part["total_sanctioned_amount"].replace(0, np.nan)
    part["disbursed_to_sanctioned_ratio"] = (
        part["total_disbursed_amount"] / denominator
    )

    # 3. Clean infinities
    part["disbursed_to_sanctioned_ratio"] = (
        part["disbursed_to_sanctioned_ratio"]
        .replace([np.inf, -np.inf], np.nan)
    )

    # 4. Cap remaining non-missing values at the training 99th percentile
    part["disbursed_to_sanctioned_ratio"] = (
        part["disbursed_to_sanctioned_ratio"]
        .clip(upper=ratio_cap)
    )

    # Keep disbursed amount, drop sanctioned amount
    drop_cols = [
        "pri_sanctioned_amount",
        "total_sanctioned_amount"
    ]

    part.drop(
        columns=[c for c in drop_cols if c in part.columns],
        inplace=True
    )

    return part

train_df = resolve_collinearity(train_df, RATIO_CAP)
val_df = resolve_collinearity(val_df, RATIO_CAP)

if heldout_test_df is not None:
    heldout_test_df = resolve_collinearity(heldout_test_df, RATIO_CAP)

print(
    "Account-age and credit-history length kept:",
    "average_acct_age_months" in train_df.columns,
    "credit_history_length_months" in train_df.columns
)

print(f"Ratio capped at 99th percentile: {RATIO_CAP:.6f}")
print("\nDisbursed/sanctioned ratio summary:")
print(train_df["disbursed_to_sanctioned_ratio"].describe())
print("\nSanctioned amount is zero (flag sum):", train_df["sanctioned_amount_is_zero"].sum())
```

```python
# 4. Bureau features
# Train-only learning: inquiry cap is calculated ONLY from train_df.

INQUIRY_CAP = train_df["no_of_inquiries"].quantile(0.99)

NEW_TO_CREDIT_MONTHS = 12


def add_bureau_features(part, inquiry_cap):
    part = part.copy()

    # No bureau history
    part["no_bureau_history"] = (
        part["perform_cns_score"] == 0
    ).astype(int)

    # Keep raw inquiry count and add capped version
    part["no_of_inquiries_capped"] = (
        part["no_of_inquiries"]
        .clip(upper=inquiry_cap)
    )

    # Any delinquency
    part["any_delinquency"] = (
        (part["pri_overdue_accts"].fillna(0) > 0)
        |
        (part["delinquent_accts_in_last_six_months"].fillna(0) > 0)
    ).astype(int)

    # New-to-credit
    part["new_to_credit"] = (
        part["credit_history_length_months"].fillna(0)
        <= NEW_TO_CREDIT_MONTHS
    ).astype(int)

    return part


train_df = add_bureau_features(train_df, INQUIRY_CAP)
val_df = add_bureau_features(val_df, INQUIRY_CAP)

if heldout_test_df is not None:
    heldout_test_df = add_bureau_features(
        heldout_test_df,
        INQUIRY_CAP
    )

preprocessing_objects = {
    "inquiry_cap_99pct": float(INQUIRY_CAP),
    "new_to_credit_threshold_months": NEW_TO_CREDIT_MONTHS
}

print("Training-only inquiry cap:", INQUIRY_CAP)

print("\nBureau features created:")
print([
    "no_bureau_history",
    "no_of_inquiries_capped",
    "any_delinquency",
    "new_to_credit"
])
```

```python
# 5. Loan features
# Decision for negative balances: CLIP AT ZERO.

def add_loan_features(part):
    part = part.copy()

    # LTV > 80% indicator
    part["ltv_over_80"] = (
        part["ltv"] > 80
    ).astype(int)

    # Loan-to-asset ratio
    asset_cost = part["asset_cost"].replace(0, np.nan)

    part["loan_to_asset_ratio"] = (
        part["disbursed_amount"] / asset_cost
    )

    part["loan_to_asset_ratio"] = (
        part["loan_to_asset_ratio"]
        .replace([np.inf, -np.inf], np.nan)
        .fillna(0)
    )

    # Clip negative balances
    balance_cols = [
        "pri_current_balance",
        "total_current_balance"
    ]

    for col in balance_cols:
        if col in part.columns:
            negative_count = (
                part[col].fillna(0) < 0
            ).sum()

            if negative_count > 0:
                print(
                    f"{col}: clipping {negative_count} "
                    f"negative values to zero"
                )

            part[col] = part[col].clip(lower=0)

    return part


train_df = add_loan_features(train_df)
val_df = add_loan_features(val_df)

if heldout_test_df is not None:
    heldout_test_df = add_loan_features(heldout_test_df)

# Amount columns to log-transform ONLY for Logistic Regression / MLP
LOG_AMOUNT_COLS = [
    "disbursed_amount",
    "asset_cost",
    "pri_current_balance",
    "total_current_balance",
    "total_disbursed_amount",
    "total_instal_amt",
    "pri_disbursed_amount",
    "pri_instal_amt"
]

LOG_AMOUNT_COLS = [
    c for c in LOG_AMOUNT_COLS
    if c in train_df.columns
]

preprocessing_objects["log_amount_cols"] = LOG_AMOUNT_COLS
preprocessing_objects["negative_balance_decision"] = "clip_at_zero"

print("\nLog-transform columns:")
print(LOG_AMOUNT_COLS)
```

```python
# 6. Demographic and document features

def add_demographic_features(part):
    part = part.copy()

    # Age bands - useful especially for linear models
    part["age_group"] = pd.cut(
        part["age_at_disbursal"],
        bins=[0, 25, 35, 45, 55, 65, 80, np.inf],
        labels=[
            "18-25",
            "26-35",
            "36-45",
            "46-55",
            "56-65",
            "66-80",
            "80+"
        ],
        include_lowest=True
    )

    # Employment type
    part["employment_type"] = (
        part["employment_type"]
        .fillna("Unknown")
        .astype(str)
    )

    # Documents provided
    document_cols = [
        "aadhar_flag",
        "pan_flag",
        "voterid_flag",
        "driving_flag",
        "passport_flag"
    ]

    document_cols = [
        c for c in document_cols
        if c in part.columns
    ]

    part["documents_provided"] = (
        part[document_cols]
        .fillna(0)
        .sum(axis=1)
    )

    return part


train_df = add_demographic_features(train_df)
val_df = add_demographic_features(val_df)

if heldout_test_df is not None:
    heldout_test_df = add_demographic_features(heldout_test_df)

print("Demographic/document features created.")

display(
    train_df[
        [
            "age_at_disbursal",
            "age_group",
            "employment_type",
            "documents_provided"
        ]
    ].head()
)
```

```python
# 7. Encode categoricals - TRAIN ONLY
#
# We use:
#   1. Rare-category grouping
#   2. Frequency encoding
#
# This avoids target leakage, so OOF target encoding is not required.

HIGH_CARD_COLS = [
    "branch_id",
    "supplier_id",
    "manufacturer_id"
]

RARE_MIN_COUNT = 50


def fit_frequency_encoders(train_part):
    rare_categories = {}
    frequency_maps = {}

    for col in HIGH_CARD_COLS:

        if col not in train_part.columns:
            continue

        s = train_part[col].astype(str)

        counts = s.value_counts()

        # Categories with fewer than minimum observations
        rare_categories[col] = set(
            counts[counts < RARE_MIN_COUNT].index
        )

        grouped = s.where(
            ~s.isin(rare_categories[col]),
            "Other"
        )

        # Frequency mapping learned ONLY from train
        frequency_maps[col] = (
            grouped.value_counts(normalize=True)
            .to_dict()
        )

    return rare_categories, frequency_maps


def apply_frequency_encoders(
    part,
    rare_categories,
    frequency_maps
):
    part = part.copy()

    for col in HIGH_CARD_COLS:

        if col not in part.columns:
            continue

        s = part[col].astype(str)

        # Map rare/unseen categories to Other
        grouped = s.where(
            ~s.isin(rare_categories[col]),
            "Other"
        )

        # Frequency encoding
        part[col + "_freq"] = (
            grouped.map(frequency_maps[col])
            .fillna(0)
        )

        # Drop original high-cardinality column
        part.drop(columns=[col], inplace=True)

    return part


# FIT ONLY ON TRAIN
rare_categories, frequency_maps = fit_frequency_encoders(train_df)

# APPLY TO all datasets
train_df = apply_frequency_encoders(
    train_df,
    rare_categories,
    frequency_maps
)

val_df = apply_frequency_encoders(
    val_df,
    rare_categories,
    frequency_maps
)

if heldout_test_df is not None:
    heldout_test_df = apply_frequency_encoders(
        heldout_test_df,
        rare_categories,
        frequency_maps
    )

preprocessing_objects["rare_categories"] = rare_categories
preprocessing_objects["frequency_maps"] = frequency_maps

print("Frequency-encoded columns:")
print([
    c for c in train_df.columns
    if c.endswith("_freq")
])
```

```python
# 8. Scale and assemble model inputs
#
# Logistic Regression + MLP:
#   - log-transform skewed amount features
#   - one-hot categoricals
#   - StandardScaler
#
# XGBoost:
#   - no log transformation
#   - one-hot categoricals
#   - no scaling
#
# All preprocessing objects are FIT ONLY ON TRAIN.

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer


TARGET = "loan_default"


def create_model_dataframe(part, apply_log=False):
    part = part.copy()

    # Remove target
    if TARGET in part.columns:
        part = part.drop(columns=[TARGET])

    # Dates are replaced by age, so don't feed raw dates
    drop_cols = [
        "date_of_birth",
        "disbursaldate",
        "perform_cns_score_description"
    ]

    part.drop(
        columns=[c for c in drop_cols if c in part.columns],
        inplace=True
    )

    # Log transforms ONLY for Logistic/MLP
    if apply_log:
        for col in LOG_AMOUNT_COLS:

            if col in part.columns:
                values = part[col].clip(lower=0)

                part[col + "_log1p"] = np.log1p(values)

                # Replace original with log version
                part.drop(columns=[col], inplace=True)

    return part


# -----------------------------
# Create model-specific frames
# -----------------------------

X_train_lr = create_model_dataframe(train_df, apply_log=True)
X_val_lr = create_model_dataframe(val_df, apply_log=True)

X_train_xgb = create_model_dataframe(train_df, apply_log=False)
X_val_xgb = create_model_dataframe(val_df, apply_log=False)

y_train = train_df[TARGET].copy()
y_val = val_df[TARGET].copy()


# Internal held-out test, if using 70/15/15
if heldout_test_df is not None:

    X_test_lr = create_model_dataframe(
        heldout_test_df,
        apply_log=True
    )

    X_test_xgb = create_model_dataframe(
        heldout_test_df,
        apply_log=False
    )

    y_test = heldout_test_df[TARGET].copy()


# -----------------------------
# Categorical columns
# -----------------------------

categorical_cols = [
    c for c in [
        "employment_type",
        "age_group",
        "state_id"
    ]
    if c in X_train_lr.columns
]


# All remaining columns are numeric
numeric_cols_lr = [
    c for c in X_train_lr.columns
    if c not in categorical_cols
]

numeric_cols_xgb = [
    c for c in X_train_xgb.columns
    if c not in categorical_cols
]


# -----------------------------
# Logistic / MLP preprocessor
# -----------------------------

linear_preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            Pipeline([
                ("imputer", SimpleImputer(strategy="median")),
                ("scaler", StandardScaler())
            ]),
            numeric_cols_lr
        ),
        (
            "cat",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            ),
            categorical_cols
        )
    ],
    remainder="drop"
)


# -----------------------------
# XGBoost preprocessor
# -----------------------------

xgb_preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            SimpleImputer(strategy="median"),
            numeric_cols_xgb
        ),
        (
            "cat",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            ),
            categorical_cols
        )
    ],
    remainder="drop"
)


# FIT ONLY ON TRAIN
X_train_lr_processed = linear_preprocessor.fit_transform(X_train_lr)

X_val_lr_processed = linear_preprocessor.transform(X_val_lr)

X_train_xgb_processed = xgb_preprocessor.fit_transform(X_train_xgb)

X_val_xgb_processed = xgb_preprocessor.transform(X_val_xgb)


# Internal test
if heldout_test_df is not None:

    X_test_lr_processed = linear_preprocessor.transform(
        X_test_lr
    )

    X_test_xgb_processed = xgb_preprocessor.transform(
        X_test_xgb
    )


# Save feature names
lr_feature_names = (
    linear_preprocessor.get_feature_names_out()
)

xgb_feature_names = (
    xgb_preprocessor.get_feature_names_out()
)

pd.DataFrame({
    "feature": lr_feature_names
}).to_csv(
    "outputs/logistic_mlp_feature_list.csv",
    index=False
)

pd.DataFrame({
    "feature": xgb_feature_names
}).to_csv(
    "outputs/xgboost_feature_list.csv",
    index=False
)


# Save everything needed later
preprocessing_objects["linear_preprocessor"] = linear_preprocessor
preprocessing_objects["xgb_preprocessor"] = xgb_preprocessor
preprocessing_objects["linear_feature_names"] = lr_feature_names.tolist()
preprocessing_objects["xgb_feature_names"] = xgb_feature_names.tolist()

joblib.dump(
    preprocessing_objects,
    "outputs/preprocessing_objects.joblib"
)

print("Logistic/MLP train shape:", X_train_lr_processed.shape)
print("Logistic/MLP validation shape:", X_val_lr_processed.shape)

print("XGBoost train shape:", X_train_xgb_processed.shape)
print("XGBoost validation shape:", X_val_xgb_processed.shape)
```

```python
# 9. Final sanity checks

def check_processed_data(name, X, y=None):
    print(f"\n===== {name} =====")

    print("Shape:", X.shape)

    print(
        "NaN values:",
        np.isnan(X).sum()
    )

    print(
        "Infinite values:",
        np.isinf(X).sum()
    )

    if y is not None:
        print(
            "Default rate:",
            f"{y.mean():.4%}"
        )


# Check processed datasets
check_processed_data(
    "Logistic / MLP Train",
    X_train_lr_processed,
    y_train
)

check_processed_data(
    "Logistic / MLP Validation",
    X_val_lr_processed,
    y_val
)

check_processed_data(
    "XGBoost Train",
    X_train_xgb_processed,
    y_train
)

check_processed_data(
    "XGBoost Validation",
    X_val_xgb_processed,
    y_val
)


# Internal 70/15/15 test
if heldout_test_df is not None:

    check_processed_data(
        "Logistic / MLP Held-out Test",
        X_test_lr_processed,
        y_test
    )

    check_processed_data(
        "XGBoost Held-out Test",
        X_test_xgb_processed,
        y_test
    )

    print("\nFeature consistency:")
    print(
        "LR/MLP:",
        X_train_lr_processed.shape[1]
        == X_val_lr_processed.shape[1]
        == X_test_lr_processed.shape[1]
    )

    print(
        "XGBoost:",
        X_train_xgb_processed.shape[1]
        == X_val_xgb_processed.shape[1]
        == X_test_xgb_processed.shape[1]
    )


# Default-rate consistency
print("\nDefault-rate comparison:")

rates = {
    "Train": y_train.mean(),
    "Validation": y_val.mean()
}

if heldout_test_df is not None:
    rates["Held-out Test"] = y_test.mean()

for name, rate in rates.items():
    print(f"{name}: {rate:.4%}")


print("\nSaved preprocessing objects:")
print("outputs/preprocessing_objects.joblib")

print("\nSaved feature lists:")
print("outputs/logistic_mlp_feature_list.csv")
print("outputs/xgboost_feature_list.csv")
```

```python
# ============================================================
# 5.1 Evaluation setup
# ============================================================

import os
import random
import json
import joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.metrics import (
    roc_auc_score,
    roc_curve,
    recall_score,
    confusion_matrix,
    classification_report
)

os.makedirs("outputs", exist_ok=True)

RANDOM_STATE = 42
np.random.seed(RANDOM_STATE)
random.seed(RANDOM_STATE)


def ks_statistic(y_true, y_prob):
    """
    KS = maximum(TPR - FPR)
    """
    fpr, tpr, thresholds = roc_curve(y_true, y_prob)

    ks_values = tpr - fpr
    max_idx = np.argmax(ks_values)

    return (
        ks_values[max_idx],
        thresholds[max_idx]
    )


def evaluate_predictions(
    y_true,
    y_prob,
    threshold=None,
    choose_threshold=False
):
    """
    Common evaluation function for all models.

    Metrics:
    - ROC-AUC
    - KS
    - Recall at chosen threshold
    """

    auc = roc_auc_score(y_true, y_prob)

    ks, ks_threshold = ks_statistic(
        y_true,
        y_prob
    )

    if choose_threshold or threshold is None:
        threshold = ks_threshold

    y_pred = (
        np.asarray(y_prob) >= threshold
    ).astype(int)

    recall = recall_score(
        y_true,
        y_pred,
        zero_division=0
    )

    return {
        "AUC": auc,
        "KS": ks,
        "Recall": recall,
        "Threshold": threshold
    }


# Prepare labels
y_train = train_df["loan_default"].astype(int).values
y_val = val_df["loan_default"].astype(int).values

print("Evaluation function created.")

print("\nTraining class distribution:")
print(pd.Series(y_train).value_counts())

NEGATIVES = (y_train == 0).sum()
POSITIVES = (y_train == 1).sum()

POS_WEIGHT = NEGATIVES / POSITIVES

print("\nNegative / Positive ratio:", POS_WEIGHT)
```

```python
# ============================================================
# 5.3 XGBoost
# ============================================================

from xgboost import XGBClassifier

# Fixed seed
rng = np.random.default_rng(RANDOM_STATE)

N_SEARCH = 20

xgb_results = []

best_xgb = None
best_xgb_result = None
best_xgb_params = None


for i in range(N_SEARCH):

    params = {
        "max_depth": int(
            rng.integers(3, 7)
        ),

        "learning_rate": float(
            rng.uniform(0.02, 0.15)
        ),

        "n_estimators": int(
            rng.integers(300, 1201)
        ),

        "subsample": float(
            rng.uniform(0.70, 0.95)
        ),

        "colsample_bytree": float(
            rng.uniform(0.70, 0.95)
        )
    }

    model = XGBClassifier(
        objective="binary:logistic",

        max_depth=params["max_depth"],
        learning_rate=params["learning_rate"],
        n_estimators=params["n_estimators"],
        subsample=params["subsample"],
        colsample_bytree=params["colsample_bytree"],

        scale_pos_weight=POS_WEIGHT,

        eval_metric="auc",

        tree_method="hist",

        random_state=RANDOM_STATE + i,
        n_jobs=-1,

        early_stopping_rounds=50
    )

    model.fit(
        X_train_xgb_processed,
        y_train,

        eval_set=[
            (
                X_val_xgb_processed,
                y_val
            )
        ],

        verbose=False
    )

    val_prob = model.predict_proba(
        X_val_xgb_processed
    )[:, 1]

    result = evaluate_predictions(
        y_val,
        val_prob,
        choose_threshold=True
    )

    result.update(params)
    result["search_iteration"] = i + 1

    # Actual number of trees used after early stopping
    result["best_iteration"] = (
        model.best_iteration
        if hasattr(model, "best_iteration")
        else np.nan
    )

    xgb_results.append(result)

    if (
        best_xgb_result is None
        or result["AUC"] > best_xgb_result["AUC"]
    ):
        best_xgb = model
        best_xgb_result = result
        best_xgb_params = params


xgb_results_df = pd.DataFrame(
    xgb_results
)

print("Top XGBoost configurations:")
display(
    xgb_results_df.sort_values(
        "AUC",
        ascending=False
    ).head(10)
)

print("\nBest XGBoost parameters:")
print(best_xgb_params)

print("\nBest XGBoost validation result:")
print(best_xgb_result)


# ------------------------------------------------------------
# Feature importance
# ------------------------------------------------------------

xgb_feature_names = np.array(
    preprocessing_objects["xgb_feature_names"]
)

xgb_importance_df = pd.DataFrame({
    "feature": xgb_feature_names,
    "importance": best_xgb.feature_importances_
})

xgb_importance_df = xgb_importance_df.sort_values(
    "importance",
    ascending=False
)

print("\nTop XGBoost features:")
display(
    xgb_importance_df.head(20)
)

xgb_importance_df.to_csv(
    "outputs/xgboost_feature_importance.csv",
    index=False
)


# Save XGBoost model
joblib.dump(
    best_xgb,
    "outputs/xgboost_model.joblib"
)
```

```python
# ============================================================
# 5.4 PyTorch MLP
# ============================================================

import torch
import torch.nn as nn
from torch.utils.data import TensorDataset, DataLoader


# ------------------------------------------------------------
# Reproducibility
# ------------------------------------------------------------

torch.manual_seed(RANDOM_STATE)

if torch.cuda.is_available():
    torch.cuda.manual_seed_all(RANDOM_STATE)


device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("Using device:", device)


# ------------------------------------------------------------
# Convert data
# ------------------------------------------------------------

X_train_mlp = np.asarray(
    X_train_lr_processed,
    dtype=np.float32
)

X_val_mlp = np.asarray(
    X_val_lr_processed,
    dtype=np.float32
)

y_train_mlp = np.asarray(
    y_train,
    dtype=np.float32
)

y_val_mlp = np.asarray(
    y_val,
    dtype=np.float32
)


# ------------------------------------------------------------
# Dataset
# ------------------------------------------------------------

train_dataset = TensorDataset(
    torch.tensor(X_train_mlp),
    torch.tensor(y_train_mlp)
)

val_dataset = TensorDataset(
    torch.tensor(X_val_mlp),
    torch.tensor(y_val_mlp)
)

train_loader = DataLoader(
    train_dataset,
    batch_size=512,
    shuffle=True
)

val_loader = DataLoader(
    val_dataset,
    batch_size=512,
    shuffle=False
)


# ------------------------------------------------------------
# Model architecture
# 128 -> 64 -> 1
# ------------------------------------------------------------

class MLP(nn.Module):

    def __init__(
        self,
        input_dim,
        dropout=0.30
    ):
        super().__init__()

        self.network = nn.Sequential(

            nn.Linear(
                input_dim,
                128
            ),

            nn.ReLU(),

            nn.Dropout(dropout),

            nn.Linear(
                128,
                64
            ),

            nn.ReLU(),

            nn.Dropout(dropout),

            nn.Linear(
                64,
                1
            )
        )

    def forward(self, x):
        return self.network(x).squeeze(1)


input_dim = X_train_mlp.shape[1]

mlp = MLP(
    input_dim=input_dim,
    dropout=0.30
).to(device)


# ------------------------------------------------------------
# Loss and optimizer
# ------------------------------------------------------------

pos_weight = torch.tensor(
    [POS_WEIGHT],
    dtype=torch.float32,
    device=device
)

criterion = nn.BCEWithLogitsLoss(
    pos_weight=pos_weight
)

optimizer = torch.optim.Adam(
    mlp.parameters(),
    lr=0.001
)


# ------------------------------------------------------------
# Training with early stopping on validation AUC
# ------------------------------------------------------------

MAX_EPOCHS = 100
PATIENCE = 10

best_val_auc = -np.inf
best_epoch = 0
patience_counter = 0

best_state = None

train_losses = []
val_losses = []
val_aucs = []


for epoch in range(MAX_EPOCHS):

    # -------------------------
    # Training
    # -------------------------

    mlp.train()

    total_train_loss = 0
    train_count = 0

    for X_batch, y_batch in train_loader:

        X_batch = X_batch.to(device)
        y_batch = y_batch.to(device)

        optimizer.zero_grad()

        logits = mlp(X_batch)

        loss = criterion(
            logits,
            y_batch
        )

        loss.backward()

        optimizer.step()

        total_train_loss += (
            loss.item()
            * X_batch.size(0)
        )

        train_count += X_batch.size(0)

    epoch_train_loss = (
        total_train_loss
        / train_count
    )


    # -------------------------
    # Validation
    # -------------------------

    mlp.eval()

    total_val_loss = 0
    val_count = 0

    val_probs = []

    with torch.no_grad():

        for X_batch, y_batch in val_loader:

            X_batch = X_batch.to(device)
            y_batch = y_batch.to(device)

            logits = mlp(X_batch)

            loss = criterion(
                logits,
                y_batch
            )

            probs = torch.sigmoid(
                logits
            )

            total_val_loss += (
                loss.item()
                * X_batch.size(0)
            )

            val_count += X_batch.size(0)

            val_probs.extend(
                probs.cpu().numpy()
            )

    epoch_val_loss = (
        total_val_loss
        / val_count
    )

    epoch_val_auc = roc_auc_score(
        y_val,
        val_probs
    )

    train_losses.append(
        epoch_train_loss
    )

    val_losses.append(
        epoch_val_loss
    )

    val_aucs.append(
        epoch_val_auc
    )


    print(
        f"Epoch {epoch + 1:03d} | "
        f"Train Loss: {epoch_train_loss:.4f} | "
        f"Val Loss: {epoch_val_loss:.4f} | "
        f"Val AUC: {epoch_val_auc:.4f}"
    )


    # -------------------------
    # Early stopping
    # -------------------------

    if epoch_val_auc > best_val_auc:

        best_val_auc = epoch_val_auc
        best_epoch = epoch + 1
        patience_counter = 0

        best_state = {
            k: v.cpu().clone()
            for k, v in mlp.state_dict().items()
        }

    else:

        patience_counter += 1

        if patience_counter >= PATIENCE:

            print(
                f"\nEarly stopping at epoch "
                f"{epoch + 1}"
            )

            break


# Restore best weights
mlp.load_state_dict(best_state)

mlp.to(device)

print("\nBest MLP epoch:", best_epoch)
print("Best validation AUC:", best_val_auc)


# ------------------------------------------------------------
# Final validation probabilities
# ------------------------------------------------------------

mlp.eval()

with torch.no_grad():

    val_logits = mlp(
        torch.tensor(
            X_val_mlp,
            dtype=torch.float32,
            device=device
        )
    )

    mlp_val_prob = (
        torch.sigmoid(
            val_logits
        )
        .cpu()
        .numpy()
    )


mlp_result = evaluate_predictions(
    y_val,
    mlp_val_prob,
    choose_threshold=True
)

print("\nMLP validation result:")
print(mlp_result)


# ------------------------------------------------------------
# Training vs validation loss
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

plt.plot(
    train_losses,
    label="Training Loss"
)

plt.plot(
    val_losses,
    label="Validation Loss"
)

plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("MLP Training vs Validation Loss")
plt.legend()

plt.tight_layout()

plt.savefig(
    "outputs/mlp_training_validation_loss.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# Save MLP
torch.save(
    {
        "model_state_dict": mlp.state_dict(),
        "input_dim": input_dim,
        "dropout": 0.30,
        "best_epoch": best_epoch
    },
    "outputs/mlp_model.pt"
)
```

```python
# ============================================================
# 5.5 Probability calibration using Platt scaling
# ============================================================

from sklearn.linear_model import LogisticRegression
from sklearn.calibration import calibration_curve
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import joblib

# ------------------------------------------------------------
# First compare the three raw validation results
# ------------------------------------------------------------

lr_val_prob = best_lr.predict_proba(
    X_val_lr_processed
)[:, 1]

xgb_val_prob = best_xgb.predict_proba(
    X_val_xgb_processed
)[:, 1]

lr_validation = evaluate_predictions(
    y_val,
    lr_val_prob,
    choose_threshold=True
)

xgb_validation = evaluate_predictions(
    y_val,
    xgb_val_prob,
    choose_threshold=True
)

mlp_validation = evaluate_predictions(
    y_val,
    mlp_val_prob,
    choose_threshold=True
)

validation_comparison = pd.DataFrame([
    {
        "Model": "Logistic Regression",
        **lr_validation
    },
    {
        "Model": "XGBoost",
        **xgb_validation
    },
    {
        "Model": "PyTorch MLP",
        **mlp_validation
    }
])

print("Validation comparison:")
display(validation_comparison)

# ------------------------------------------------------------
# Select best model using validation AUC
# ------------------------------------------------------------

best_model_name = (
    validation_comparison
    .sort_values(
        ["AUC", "KS"],
        ascending=False
    )
    .iloc[0]["Model"]
)

print(
    "Best model selected from validation:",
    best_model_name
)

# ------------------------------------------------------------
# Get best model validation probabilities
# ------------------------------------------------------------

model_probabilities = {
    "Logistic Regression": lr_val_prob,
    "XGBoost": xgb_val_prob,
    "PyTorch MLP": mlp_val_prob
}

best_raw_val_prob = model_probabilities[
    best_model_name
]

# ------------------------------------------------------------
# Split validation into calibration/evaluation portions
# This avoids fitting and checking calibration on exactly
# the same observations.
# ------------------------------------------------------------

calibration_prob_train, calibration_prob_eval, \
calibration_y_train, calibration_y_eval = train_test_split(
    best_raw_val_prob,
    y_val,
    test_size=0.50,
    stratify=y_val,
    random_state=RANDOM_STATE
)

# ------------------------------------------------------------
# Platt calibration
# Logistic regression on log-odds of model probability
# ------------------------------------------------------------

eps = 1e-7

calibration_prob_train_clipped = np.clip(
    calibration_prob_train,
    eps,
    1 - eps
)

logit_train = np.log(
    calibration_prob_train_clipped
    /
    (1 - calibration_prob_train_clipped)
).reshape(-1, 1)

platt_calibrator = LogisticRegression(
    C=1e6,
    max_iter=1000
)

platt_calibrator.fit(
    logit_train,
    calibration_y_train
)

def calibrate_probability(
    probabilities,
    calibrator
):
    probabilities = np.clip(
        probabilities,
        eps,
        1 - eps
    )

    logit = np.log(
        probabilities
        /
        (1 - probabilities)
    ).reshape(-1, 1)

    return calibrator.predict_proba(
        logit
    )[:, 1]

calibrated_val_prob = calibrate_probability(
    calibration_prob_eval,
    platt_calibrator
)

# ------------------------------------------------------------
# Choose final threshold AFTER calibration on validation eval
# ------------------------------------------------------------

calibrated_validation_result = evaluate_predictions(
    calibration_y_eval,
    calibrated_val_prob,
    choose_threshold=True
)

chosen_threshold = calibrated_validation_result[
    "Threshold"
]

print("\nCalibrated validation result:")
print(calibrated_validation_result)

print(
    "\nChosen final threshold:",
    chosen_threshold
)

# ------------------------------------------------------------
# Calibration curve strictly using validation eval data
# ------------------------------------------------------------

prob_true, prob_pred = calibration_curve(
    calibration_y_eval,
    calibrated_val_prob,
    n_bins=10,
    strategy="quantile"
)

plt.figure(figsize=(7, 6))

plt.plot(
    prob_pred,
    prob_true,
    marker="o",
    label="Calibrated model (Validation Eval)"
)

plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    label="Perfect calibration"
)

plt.xlabel("Mean predicted probability")
plt.ylabel("Observed default rate")
plt.title(
    f"Calibration Curve - {best_model_name} (Validation)"
)

plt.legend()
plt.tight_layout()

plt.savefig(
    "outputs/calibration_curve.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# Save calibrator
joblib.dump(
    platt_calibrator,
    "outputs/platt_calibrator.joblib"
)

joblib.dump(
    chosen_threshold,
    "outputs/chosen_threshold.joblib"
)
```

```python
# ============================================================
# 6.2 Choose PD band cutoffs using VALIDATION ONLY
# ============================================================

low_cutoff = np.quantile(
    val_pd,
    0.40
)

high_cutoff = np.quantile(
    val_pd,
    0.80
)

band_cutoffs = {
    "low_cutoff": float(low_cutoff),
    "high_cutoff": float(high_cutoff),
    "method": "Validation quantiles: 40% / 80%",
    "low_rule": f"PD <= {low_cutoff:.6f}",
    "medium_rule": (
        f"{low_cutoff:.6f} < PD <= {high_cutoff:.6f}"
    ),
    "high_rule": f"PD > {high_cutoff:.6f}"
}

print("Frozen validation cutoffs:")
print(f"Low cutoff:    {low_cutoff:.6f}")
print(f"High cutoff:   {high_cutoff:.6f}")


def assign_pd_band(
    probabilities,
    low_cutoff,
    high_cutoff
):
    return pd.cut(
        probabilities,
        bins=[
            -np.inf,
            low_cutoff,
            high_cutoff,
            np.inf
        ],
        labels=[
            "Low",
            "Medium",
            "High"
        ],
        include_lowest=True
    )


val_bands = assign_pd_band(
    val_pd,
    low_cutoff,
    high_cutoff
)

if heldout_test_df is not None:

    test_bands = assign_pd_band(
        test_pd,
        low_cutoff,
        high_cutoff
    )

print("\nValidation band counts:")
print(val_bands.value_counts().sort_index())
```

```python
# ============================================================
# 6.3 Band summary table
# ============================================================

PORTFOLIO_DEFAULT_RATE = df["loan_default"].mean()


def create_band_summary(
    y_true,
    predicted_pd,
    bands,
    portfolio_default_rate
):

    data = pd.DataFrame({
        "loan_default": y_true,
        "predicted_pd": predicted_pd,
        "pd_band": bands
    })

    total_defaults = data["loan_default"].sum()

    summary = (
        data.groupby(
            "pd_band",
            observed=False
        )
        .agg(
            loans=("loan_default", "size"),
            defaults=("loan_default", "sum"),
            average_predicted_pd=("predicted_pd", "mean"),
            actual_default_rate=("loan_default", "mean")
        )
        .reset_index()
    )

    summary["portfolio_share"] = (
        summary["loans"]
        / len(data)
    )

    summary["defaults_captured"] = (
        summary["defaults"]
        / total_defaults
        if total_defaults > 0
        else 0
    )

    summary["lift"] = (
        summary["actual_default_rate"]
        / portfolio_default_rate
    )

    # Convert to percentages
    summary["portfolio_share_pct"] = (
        summary["portfolio_share"] * 100
    )

    summary["actual_default_rate_pct"] = (
        summary["actual_default_rate"] * 100
    )

    summary["average_predicted_pd_pct"] = (
        summary["average_predicted_pd"] * 100
    )

    summary["defaults_captured_pct"] = (
        summary["defaults_captured"] * 100
    )

    summary["lift"] = summary["lift"].round(3)

    return summary[
        [
            "pd_band",
            "loans",
            "portfolio_share_pct",
            "defaults",
            "actual_default_rate_pct",
            "average_predicted_pd_pct",
            "defaults_captured_pct",
            "lift"
        ]
    ]


validation_band_summary = create_band_summary(
    y_val,
    val_pd,
    val_bands,
    PORTFOLIO_DEFAULT_RATE
)

print("VALIDATION BAND SUMMARY")
display(validation_band_summary)


if heldout_test_df is not None:

    test_band_summary = create_band_summary(
        y_test,
        test_pd,
        test_bands,
        PORTFOLIO_DEFAULT_RATE
    )

    print("\nHELD-OUT TEST BAND SUMMARY")
    display(test_band_summary)

    validation_band_summary.to_csv(
        "outputs/validation_pd_bands.csv",
        index=False
    )

    test_band_summary.to_csv(
        "outputs/test_pd_bands.csv",
        index=False
    )
```

```python
# ============================================================
# 7.2 Rejection sweep - VALIDATION
# ============================================================

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

os.makedirs("outputs", exist_ok=True)


def rejection_sweep(
    y_true,
    predicted_pd,
    rejection_percentages
):
    """
    Reject the riskiest X% of applicants.

    Returns:
    - defaults avoided
    - approvals lost
    - remaining portfolio default rate
    - good loans rejected
    """

    data = pd.DataFrame({
        "loan_default": np.asarray(y_true),
        "predicted_pd": np.asarray(predicted_pd)
    })

    # Highest PD first
    data = data.sort_values(
        "predicted_pd",
        ascending=False
    ).reset_index(drop=True)

    total_loans = len(data)
    total_defaults = data["loan_default"].sum()
    total_good_loans = (
        total_loans - total_defaults
    )

    results = []

    for pct in rejection_percentages:

        # Number of loans to reject
        n_reject = int(
            round(total_loans * pct)
        )

        rejected = data.iloc[:n_reject]
        approved = data.iloc[n_reject:]

        defaults_rejected = (
            rejected["loan_default"].sum()
        )

        good_loans_rejected = (
            n_reject - defaults_rejected
        )

        defaults_avoided_pct = (
            defaults_rejected / total_defaults * 100
            if total_defaults > 0
            else np.nan
        )

        approvals_lost_pct = (
            n_reject / total_loans * 100
        )

        good_loans_rejected_pct = (
            good_loans_rejected / total_good_loans * 100
            if total_good_loans > 0
            else np.nan
        )

        remaining_default_rate = (
            approved["loan_default"].mean()
            if len(approved) > 0
            else np.nan
        )

        results.append({
            "rejection_pct": pct * 100,
            "loans_rejected": n_reject,
            "defaults_rejected": defaults_rejected,
            "defaults_avoided_pct": defaults_avoided_pct,
            "approvals_lost_pct": approvals_lost_pct,
            "approved_loans": len(approved),
            "approved_default_rate_pct":
                remaining_default_rate * 100,
            "good_loans_rejected_pct":
                good_loans_rejected_pct
        })

    return pd.DataFrame(results)


rejection_percentages = np.arange(
    0.05,
    0.51,
    0.05
)

validation_sweep = rejection_sweep(
    y_val,
    val_pd,
    rejection_percentages
)

print("Validation rejection sweep:")
display(validation_sweep)

validation_sweep.to_csv(
    "outputs/validation_rejection_sweep.csv",
    index=False
)
```

```python
# ============================================================
# 7.5 Optional cost / value analysis
# ============================================================

# These are ILLUSTRATIVE assumptions only.
# Change them to perform sensitivity analysis.
#
# L = loss associated with a default
# P = profit from a good approved loan
#
# Only the ratio L/P matters for comparing operating points.

LOSS_PER_DEFAULT = 5.0
PROFIT_PER_GOOD_LOAN = 1.0


def calculate_net_value(
    sweep_df,
    y_true,
    loss_per_default,
    profit_per_good
):

    data = pd.DataFrame({
        "loan_default": np.asarray(y_true),
        "predicted_pd": val_pd
    }).sort_values(
        "predicted_pd",
        ascending=False
    ).reset_index(drop=True)

    total_defaults = data["loan_default"].sum()

    values = []

    for _, row in sweep_df.iterrows():

        n_reject = int(
            row["loans_rejected"]
        )

        rejected = data.iloc[:n_reject]
        approved = data.iloc[n_reject:]

        approved_defaults = (
            approved["loan_default"].sum()
        )

        approved_good = (
            len(approved)
            - approved_defaults
        )

        # Simplified illustrative value:
        # profit from good approved loans
        # minus loss on defaults in approved portfolio
        net_value = (
            approved_good * profit_per_good
            - approved_defaults * loss_per_default
        )

        values.append(net_value)

    result = sweep_df.copy()

    result["illustrative_net_value"] = values

    return result


validation_cost_sweep = calculate_net_value(
    validation_sweep,
    y_val,
    LOSS_PER_DEFAULT,
    PROFIT_PER_GOOD_LOAN
)

print(
    "Illustrative cost/value analysis "
    f"(Loss={LOSS_PER_DEFAULT}, "
    f"Profit={PROFIT_PER_GOOD_LOAN})"
)

display(validation_cost_sweep)

best_cost_row = (
    validation_cost_sweep
    .loc[
        validation_cost_sweep[
            "illustrative_net_value"
        ].idxmax()
    ]
)

print("\nIllustrative value-maximizing point:")
display(
    best_cost_row.to_frame().T
)

validation_cost_sweep.to_csv(
    "outputs/validation_cost_sweep.csv",
    index=False
)
```

```python
# ============================================================
# 7.6 Validate frozen operating point on TEST
# ============================================================

if heldout_test_df is not None:

    # IMPORTANT:
    # Use the validation cutoff directly.
    # Do NOT calculate a new test quantile.

    test_rejected = (
        test_pd > OPERATING_CUTOFF
    )

    test_approved = ~test_rejected

    test_total_loans = len(y_test)
    test_total_defaults = y_test.sum()

    test_defaults_rejected = (
        y_test[test_rejected].sum()
    )

    test_defaults_avoided_pct = (
        test_defaults_rejected
        / test_total_defaults
        * 100
    )

    test_approved_default_rate = (
        y_test[test_approved].mean()
        * 100
    )

    test_approval_loss_pct = (
        test_rejected.mean()
        * 100
    )

    test_total_good = (
        test_total_loans
        - test_total_defaults
    )

    test_good_rejected = (
        test_rejected.sum()
        - test_defaults_rejected
    )

    test_good_rejected_pct = (
        test_good_rejected
        / test_total_good
        * 100
    )

    test_operating_result = pd.DataFrame([{
        "PD_cutoff": OPERATING_CUTOFF,
        "applicants_rejected_pct":
            test_approval_loss_pct,
        "defaults_avoided_pct":
            test_defaults_avoided_pct,
        "approved_default_rate_pct":
            test_approved_default_rate,
        "good_loans_rejected_pct":
            test_good_rejected_pct,
        "approved_loans":
            test_approved.sum(),
        "rejected_loans":
            test_rejected.sum()
    }])

    print("TEST OPERATING POINT:")
    display(test_operating_result)


    # --------------------------------------------------------
    # Full test sweep
    # --------------------------------------------------------

    test_sweep = rejection_sweep(
        y_test,
        test_pd,
        rejection_percentages
    )

    print("\nTest rejection sweep:")
    display(test_sweep)

    test_sweep.to_csv(
        "outputs/test_rejection_sweep.csv",
        index=False
    )


    # --------------------------------------------------------
    # Test trade-off curve
    # --------------------------------------------------------

    plt.figure(figsize=(8, 6))

    plt.plot(
        test_sweep["approvals_lost_pct"],
        test_sweep["defaults_avoided_pct"],
        marker="o",
        label="Model-based rejection"
    )

    plt.plot(
        [0, 50],
        [0, 50],
        linestyle="--",
        label="Random rejection"
    )

    plt.scatter(
        test_approval_loss_pct,
        test_defaults_avoided_pct,
        s=100,
        label=(
            f"Frozen cutoff "
            f"(PD > {OPERATING_CUTOFF:.3f})"
        ),
        zorder=5
    )

    plt.xlabel("% Applicants Rejected")
    plt.ylabel("% Defaults Avoided")
    plt.title("Rejection Trade-off Curve - Held-out Test")

    plt.xlim(0, 55)
    plt.ylim(0, 105)

    plt.legend()
    plt.grid(alpha=0.3)

    plt.tight_layout()

    plt.savefig(
        "outputs/test_rejection_tradeoff.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()

else:

    print(
        "No held-out test set with loan_default is available."
    )
```
